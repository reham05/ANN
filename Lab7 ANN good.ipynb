```python
import numpy as np
#Q1 a:
def logsigmoid(n):
    output = 1 / (1 + np.exp(-n))
    return output

def logsiglayer(W1,p,b1):
    n=(np.dot(W1,p))+b1
    a1=logsigmoid(n)
    return a1

#Q1 b:
def purelin(n):
    return n

def linearlayer(W2,a1,b2):
    n=(np.dot(W2,a1))+b2
    a2=purelin(n)
    return a2
```

```python
#Q2
def forwardpropagation(p, W1, b1, W2, b2):
    a0= p
    a1= logsiglayer(W1,p,b1)
    a2= linearlayer(W2,a1,b2)
    return a0, a1, a2
```

```python
#Q5
def updateparams(W1, W2, b1, b2, s1, s2, a0, a1, a2, alpha):
    W1= W1 - (alpha *s1)*np.transpose(a0)
    b1= b1-alpha*s1
    W2= W2-alpha*(np.dot(s2, a1.T))
    b2= b2-alpha*s2
    return W1, b1, W2, b2
```

```python
import numpy as np

# Initialize weights and biases
W1 = np.random.rand(2, 1)
b1 = np.random.rand(2, 1)
W2 = np.random.rand(1, 2)
b2 = np.random.rand(1, 1)

# Data points and targets
data = [
    (-2, 0.0),
    (-1.5, 0.006),
    (-1, 0.29),
    (-0.5, 0.617),
    (0, 1.0),
    (0.5, 1.38),
    (1, 1.707),
    (1.5, 1.92),
    (2, 2.0)
]

# Learning rate
alpha = 0.01  # Reduced learning rate

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Backpropagation algorithm
def backpropagatealgorithm(W1, W2, b1, b2, p, t, alpha):
    for _ in range(1000):  # Just for demonstration, normally you would use a convergence criterion
        # Forward pass
        h = np.dot(W1, np.array([[p]])).reshape(2, 1) + b1
        a = sigmoid(h)
        y = np.dot(W2, a) + b2
        E = 0.5 * (t - y) ** 2

        # Backward pass
        delta2 = -(t - y) * sigmoid_derivative(y)
        delta1 = np.dot(W2.T, delta2) * sigmoid_derivative(a)

        # Update weights and biases
        W2 -= alpha * np.dot(delta2, a.T)
        b2 -= alpha * delta2
        W1 -= alpha * np.dot(delta1, np.array([[p]]).T)
        b1 -= alpha * delta1

# Iterate through data points
for p, t in data:
    backpropagatealgorithm(W1, W2, b1, b2, p, t, alpha)

# Print converged weights and biases
print("Converged W1:")
print(W1)
print("Converged b1:")
print(b1)
print("Converged W2:")
print(W2)
print("Converged b2:")
print(b2)
```
